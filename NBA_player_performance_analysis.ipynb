```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel()
```

```python
Selected_stats =['minutes','points_pg','rebounds_pg','assists_pg','field_goal_pct','three_point_pct','effective_fg_pct','ts_pct','usage_rate','net_rating']
anaylysis_df= df[Selected_stats]
anaylysis_df.head()
```

```python
#create a scatter plot in order to visually see the correlation between the data points (scatterplot for mv points, mvrebounds, mvassists)
plt.scatter(anaylysis_df['minutes'],anaylysis_df['points_pg'])
plt.xlabel(['minutes'])
plt.ylabel(['points_pg'])
plt.title('Minutes vs points')
plt.show()

correlation= anaylysis_df['minutes'].corr(anaylysis_df['points_pg'])
print(correlation)

#- Strong + relationship
```

```python
plt.scatter(anaylysis_df['minutes'],anaylysis_df['rebounds_pg'])
plt.xlabel(['minutes'])
plt.ylabel(['rebounds_pg'])
plt.title('Minutes vs rebounds')
plt.show()

correlation= anaylysis_df['minutes'].corr(anaylysis_df['rebounds_pg'])
print(correlation)
#- moderate/ + relationship
```

```python
plt.scatter(anaylysis_df['minutes'],anaylysis_df['assists_pg'])
plt.xlabel(['minutes'])
plt.ylabel(['assists_pg'])
plt.title('Minutes vs assists')
plt.show()

Correlation= anaylysis_df['minutes'].corr(anaylysis_df['assists_pg'])
print(Correlation)
#- moderate/ + relationship
```

Through analyzing the graphs we have found a positive relationship between playing time and three productive statistics: rebounds, points and assists. The strongest correlation are between points and minutes while rebounds and assists are pretty moderate. This leads me to the conclusion that players who typically get more minutes usually get more points. Now, we do need to keep in mind that the correlation does not imply causation, because stronger performing players typically get more player times then others.
